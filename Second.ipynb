```python
Price = float(input("Enter purchase amount: Rs"))    
```

```python
if Price >= 5000:
    discount = Price * 0.20
elif Price <= 2000 and Price >=4999:
    discount = Price * 0.10
else:
    discount = 0
    
```

```python
final_amount = Price - discount
```

```python
print("Original Price: Rs",Price)
print("Discount Price: Rs", discount)
print("Final pyable Amount:Rs",final_amount)
```
